```python
df["GPA_Change"] = df["Post_Semester_GPA"] - df["Pre_Semester_GPA"]

df[["Pre_Semester_GPA", "Post_Semester_GPA", "GPA_Change"]].head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Weekly_GenAI_Hours",
    y="Post_Semester_GPA",
    alpha=0.3
)

plt.title("Weekly GenAI Usage vs Post-Semester GPA")
plt.xlabel("Weekly GenAI Hours")
plt.ylabel("Post-Semester GPA")

plt.show()
```

```python
df["Weekly_GenAI_Hours"].corr(df["Post_Semester_GPA"])
```

```python
avg_gpa = df.groupby("Year_of_Study")[
    ["Pre_Semester_GPA", "Post_Semester_GPA", "GPA_Change"]
].mean()

```

```python
avg_gpa
```

```python
avg_gpa.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Average Pre-Semester and Post-Semester GPA by Year of Study")
plt.xlabel("Year of Study")
plt.ylabel("Average GPA")
plt.xticks(rotation=0)
plt.legend(title="GPA")
plt.show()
```
